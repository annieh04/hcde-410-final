# Final Project Proposal: Does the Pro Make the Star?

Dancing with the Stars (DWTS) is a reality competition TV show currently on its 35th season. Each season, a celebrity gets partnered with a ballroom dance professional and they compete with a new dance style each week. Judges scores and fan votes are combined to determine which couple gets eliminated, and the show ends with one ultimate winner.

## Motivation & problem
Proposed project: estimate how much DWTS professional partners raise their celebrities' judges' scores and how much faster they make them improve, using Seasons 1–34.

The celebrity is the focus of the show, so fans often credit high scores to celebrity ability. But that credit mixes the pro's skill with the talent of the celebrity they were given. Separating the two is the problem. Through investing this, I am curious to learn how much of the variation in scores is explained by the pro's teaching ability, versus the celebrity's initial ability.

Not only is this interesting to fans of the show who are curious to know which Pros are the "best" at teaching, but it also provides a case study that investigates how much value the teacher adds, which can be applied to countless scenarios. The results of this project could also be used to predict which couple will win in the current and upcoming seasons based on the pro pairing.

Why they fit: Contains the weekly scores needed to evaluate performance, as well as the pro pairings.   
Ethical considerations: 
- The results of this analysis don't determine the pros proficiency or skill level. There are countless other factors that influence performance aside from the pro dancer, including the fact that judges' scores are subjective.
- Several pros only appear in one season. This provides an extremely limited dataset for these pros. I will not rank pros who have less than 4 seasons.


## Unknowns & dependencies
- There are occasional weeks when a celebrity danced with a substitute pro, which are not accounted for in the dataset. Entirely remvoing this discrepancy would require cross checking with another dataset, which may not be feasible within the given timeline.
