# Final Project Proposal: Does the Pro Make the Star?

## Overview
Dancing with the Stars (DWTS) is a reality competition TV show currently on its 35th season. Each season, a celebrity gets partnered with a ballroom dance professional and they compete with a new dance style each week. Judges scores and fan votes are combined to determine which couple gets eliminated, and the show ends with one ultimate winner.

This project aims to determine how much of a celebrity's success on DWTS comes from their professional partner. I will estimate two effects for each pro: a level effect (how much higher their celebrities score on average) and a growth effect (how much faster their celebrities improve from week to week).

### Research Questions
1. How much of the variation in judges' scores is explained by the pro, compared with the celebrity?
2. Which pros' score reliably above average (level)?
3. Which pros' improve faster over a season (growth)?

### Hypotheses
H1: The celebrity explains more of the variation in scores than the pro, but the pro still has a considerable effect.
H2: Pros differ in growth, not just level: some pros' celebrities improve noticeably faster than others'.

## Motivation & problem
Proposed project: estimate how much DWTS professional partners raise their celebrities' judges' scores and how much faster they make them improve, using Seasons 1–34.

The celebrity is the focus of the show, so fans often credit high scores to celebrity ability. But that credit mixes the pro's skill with the talent of the celebrity they were given. The celebrity's talent matters, but each routine is choreographed and taught by the pro partner, which means the pro's impact is significant.  Separating the two is the focus. Through investing this, I am curious to learn how much of the variation in scores is explained by the pro's teaching ability, versus the celebrity's initial ability.

Not only is this interesting to fans of the show who are curious to know which Pros are the "best" at teaching, but it also provides a case study that investigates how much value the teacher adds, which can be applied to countless scenarios. 

## Background
**Teacher value-added.** Separating a teacher's contribution from a student's starting ability is a well-studied problem. Chetty, Friedman and Rockoff (2014) estimate teachers' effects on test scores and show that estimates can be biased when students are not randomly assigned to teachers. This project emulates that study: each pro is a "teacher," each celebrity a "student," and the judges' score is the outcome. 
- Source: Chetty, R., Friedman, J. N., & Rockoff, J. E. (2014). Measuring the impacts of teachers I: Evaluating bias in teacher value-added estimates. American Economic Review, 104(9), 2593–2632. https://econpapers.repec.org/RePEc:aea:aecrev:v:104:y:2014:i:9:p:2593-2632

**Judges' scores depend on performance order.** Collins, McKenzie and Vaughan Williams (2019) studied DWTS and the UK's Strictly Come Dancing and found that judges give higher scores to couples who perform later in an episode. The COMAP dataset used here does not record performance order, so this is listed as a limitation.
- Source: Collins, A., McKenzie, J., & Vaughan Williams, L. (2019). When is a talent contest not a talent contest? Sequential performance bias in expert evaluation. Economics Letters, 177, 94–98. https://ideas.repec.org/a/eee/ecolet/v177y2019icp94-98.html

**The pro effect is an open question.** COMAP's 2026 Mathematical Contest in Modeling asked teams to analyze the impact of pro dancers on how well a celebrity does. However, I have not found published work that separates a pro's level effect from their growth effect. That question is the motivation for this project.

## Methodology
I’m going to use the COMAP dataset to measure how much a pro dancer affects their celebrity’s scores. I will use Python and the pandas library to clean and organize the data so that each row is one celebrity’s score in one week.

1. To make scores comparable, I will convert each week’s total judges’ score to a percent of the maximum possible score, because some weeks have 3 judges and others have 4. I will only include celebrities who competed for at least 4 weeks, since a trend based on only 2 or 3 weeks is less substantial. I will also only include pros who have 4 or more seasons.
2. I will  use a a mixed-effects model that splits each weekly score into the celebrity’s own ability, how much the pro raises their scores, and how fast the pro’s celebrities improve. The model also accounts for the fact that everyone tends to improve over a season. I chose this model because each celebrity has several weekly scores and each pro has partnered with several celebrities. It also pulls estimates for pros with only a few seasons toward the average.
3. I will present my findings with simple visualizations: a bar chart showing how much score differences come from the pro versus the celebrity and a ranked chart of pros on how high their celebrities score and how quickly they improve.
4. Finally, I will summarize which pros stand out, how large the pro’s influence really is, and possible reasons behind the results.

**Limitations.**
Producers make the pairings, so the pro effect cannot be fully separated from casting. Stronger/more popular pros tend to be paired with more athletic/entertainment background celebrities. If so, a pro's level or growth effect may partially come from the celebrities they were given. A mixed model accounts for differences between individual celebrities, but it cannot fully separate a pro's effect from celebrity traits that are linked to who gets assigned to that pro. Results should be read as contributing factors, not proof that a pro caused the improvement.
Judges' scores are subjective and affected by factors outside the data, such as the order in which couples perform.
